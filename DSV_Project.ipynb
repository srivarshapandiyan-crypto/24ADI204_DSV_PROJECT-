# WEEK 3 – The Cleaning Sprint

## Aim

To improve the quality of the dataset by identifying and handling missing values, duplicate records, inconsistent values, and numerical outliers.

## Cleaning Steps

1. Analyze missing values.
2. Handle missing values using appropriate strategies.
3. Remove duplicate records if present.
4. Check inconsistent categorical values.
5. Detect numerical outliers using the IQR method.
6. Visualize outliers using boxplots.
7. Decide whether detected extreme values are errors or genuine observations.
8. Save the cleaned dataset.

```python
#Missing values
df.isnull().sum()
```

```python
#Handling Missing Values
#For a categorical column such as festival:
if "festival" in df.columns:
    df["festival"] = df["festival"].fillna("No Festival")
```

```python
#Duplicates
print("Duplicate rows before cleaning:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
#Check Categorial(spelling/capitalization) Consistency
for column in categorical_columns:
    print("\n==============================")
    print(column)
    print("==============================")
    print(df[column].value_counts(dropna=False))
```

## Numerical Outlier Detection

```python
numeric_columns = df.select_dtypes(include=np.number).columns

outlier_report = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = ((df[column] < lower_limit) |
                (df[column] > upper_limit)).sum()

    outlier_report.append([
        column,
        Q1,
        Q3,
        IQR,
        lower_limit,
        upper_limit,
        outliers
    ])

outlier_report = pd.DataFrame(
    outlier_report,
    columns=[
        "Column",
        "Q1",
        "Q3",
        "IQR",
        "Lower Limit",
        "Upper Limit",
        "Outlier Count"
    ]
)

outlier_report
```

## Box Plot

```python
for column in numeric_columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[column])
    plt.title(f"Boxplot of {column}")
    plt.show()
```

## Cleaning Decision

The detected anomalies were examined before removal. Extreme values that represent realistic accident observations were retained, while only invalid, inconsistent, or clearly erroneous records were considered for correction or removal.

Missing values and duplicate records were handled based on the characteristics of the corresponding variables.

```python
df.to_csv("Indian_Road_Accident_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully")
```

## Correlation Analysis

Correlation analysis is used to understand the relationships between numerical variables.

A correlation value close to:

- +1 indicates a strong positive relationship.
- -1 indicates a strong negative relationship.
- 0 indicates a weak or no linear relationship.

The correlation matrix will help identify relationships between numerical variables and will also support later feature-selection and dimensionality-reduction work.

```python
correlation = df.select_dtypes(include=np.number).corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Matrix of Numerical Variables")

plt.show()
```

## Univariate Findings

1. **The distribution of accident severity shows** that minor accidents are the most common, followed by major accidents, while fatal accidents form the smallest category.

2. **The city-wise distribution shows** that accidents are distributed across eight major Indian cities, with differences in the number of recorded accidents between cities. Chennai accounts for 2,575 recorded accidents in the analyzed dataset.

3. **The most frequently observed weather condition is** clear weather, with 6,690 accidents, followed closely by rain with 6,677 accidents and fog with 6,633 accidents.

4. **The visibility distribution shows** that low visibility is the most frequently recorded condition, with 9,987 accidents, followed by high visibility with 6,690 accidents and medium visibility with 3,323 accidents.

5. **The major accident causes include** factors such as overspeeding, distraction, weather, and poor road conditions. These causes are distributed differently across the accident records.

6. **The accident distribution across hours shows** that accidents occur throughout the day, with noticeable variation across different hours. The dataset analysis also identifies evening hours as an important period for accident occurrence.

## Bivariate Findings

1. **Weather and accident severity show** that all three weather conditions contain minor, major, and fatal accidents. The differences between weather categories are relatively small, indicating that severity is not determined by weather alone.

2. **Visibility and accident severity show** that low visibility is associated with a large number of accident records. This suggests that visibility conditions are an important environmental factor to investigate further.

3. **Traffic density and accident severity show** differences in the distribution of minor, major, and fatal accidents across traffic-density categories, indicating that traffic conditions may be related to accident severity.

4. **Accident causes and severity show** that different causes are associated with different levels of accident severity. Human-related and environmental factors such as overspeeding, distraction, weather, and poor road conditions appear across the severity categories.
